```python
import ee
import sys
sys.path.append('..')

from scripts.alert_grouping_navigator import AlertGroupingNavigator
from scripts.alert_navigator_ui import AlertNavigatorUI

ee.Initialize()
```

## 2. Configuration Parameters

Set up your GEE asset and analysis parameters.

```python
# ========== CONFIGURATION PARAMETERS ==========

# GEE Asset ID containing alert data
ASSET_ID = "projects/ee-dfgm2006/assets/fires_colombia/Change_Alerts_S2"

# Area of Interest (AOI) - Define as ee.Geometry
AOI = ee.Image(ASSET_ID).geometry()

# Alert filtering parameters
MIN_ALERT_SIZE_PIXELS = 5      # Minimum pixels per alert cluster
MAX_ALERTS = 30                # Maximum number of alerts to retrieve
SORTING_METHOD = "largest_first"  # Options: largest_first, smallest_first, newest_first, oldest_first
PIXEL_SIZE = 30                # Pixel size in meters (used for area calculation)

print("Configuration set:")
print(f"  Asset: {ASSET_ID}")
print(f"  Max alerts: {MAX_ALERTS}")
print(f"  Sorting: {SORTING_METHOD}")
print(f"  Pixel size: {PIXEL_SIZE}m")
```

## 3. Load Alerts

Initialize the navigator and group alerts from your GEE asset.

```python
# Create the navigator
navigator = AlertGroupingNavigator(asset_id=ASSET_ID, pixel_size=PIXEL_SIZE)

# Convert AOI to FeatureCollection
aoi_fc = ee.FeatureCollection([ee.Feature(AOI)])

print("Grouping alerts... (this may take a few minutes)")

# Group alerts
alerts_gdf = navigator.group_alerts(
    aoi=aoi_fc,
    min_alert_size_pixels=MIN_ALERT_SIZE_PIXELS,
    max_alerts=MAX_ALERTS,
    sorting_method=SORTING_METHOD,
    grid_size=150000
)

print(f"\n✓ Successfully loaded {navigator.get_alert_count()} alerts!")
print(f"Current alert index: {navigator.get_current_index()}")
```

```python
# Create and display the interactive UI
ui = AlertNavigatorUI(navigator)
ui.display()
```
